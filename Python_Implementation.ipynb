```python
# 1. Student Attendance Analysis
# 6. Customer Purchase History
def longest_uni(a):
  maximum=0
  for i in range(len(a)):
    seen=set()
    for j in range(i,len(a)):
      if a[j] in seen:
        break
      seen.add(a[j])
      if len(seen)>maximum:
        maximum=len(seen)
  return maximum
a=list(map(int,input("Enter the input").split()))
print("Longest Unique sequence:",longest_uni(a))
```

```python
# 2. Online Shopping Price Analysis
# 4. Employee Performance Analysis

def max_sum(a):
  maximum=a[0]
  current=a[0]
  for i in range(1,len(a)):
    current=max(a[i],current+a[i])
    if current>maximum:
      maximum=current
  return maximum
a=list(map(int,input("Enter the number ").split()))
print("Maximum sum array:",max_sum(a))
```

```python
# 3. Rainwater Collection System

def trap(a):
  w=0
  for i in range(1,len(a)-1):
    l=max(a[:i])
    r=max(a[i+1:])
    level=min(l,r)
    if level>a[i]:
      w+=level-a[i]
  return w
a=list(map(int,input("Enter the input:").split()))
print("Trapped water : ",trap(a))
```

```python
# 5. Product Sales Analysis
def max_pro(a):
  maxi=a[0]
  mini=a[0]
  res=a[0]
  for i in range(1,len(a)):
    x=a[i]
    if x<0:
      maxi,mini=mini,maxi
    maxi=max(x,maxi*x)
    mini=min(x,mini*x)
    res=max(res,maxi)
  return res
a=list(map(int,input("Enter sales values: ").split()))
print("Maximum Product:",max_pro(a))
```

```python
#  7. Bank Transaction Analysis

def count_subArrays(a,target):
  count=0
  for i in range(len(a)):
    total=0
    for j in range(i,len(a)):
      total+=a[j]
      if total==target:
        count+=1
  return count
a=list(map(int,input("Enter the input: ").split()))
target=int(input("Enter the target: "))
print("Number of grp: ",count_subArrays(a,target))
```

```python
# 10. Hospital Appointment Scheduling

def merge_intervals(a):
    a.sort()
    result = []
    for start, end in a:
        if not result or start > result[-1][1]:
            result.append([start, end])
        else:
            result[-1][1] = max(result[-1][1], end)
    return result
n = int(input("Enter number of appointments: "))
a = []
for i in range(n):
    start, end = map(int, input("Enter start and end time: ").split())
    a.append([start, end])
print("Merged appointments:", merge_intervals(a))
```

```python
# 9. Network Packet Analysis

def longest_consecutive(a):
    numbers = set(a)
    maximum = 0
    for num in numbers:
        if num - 1 not in numbers:
            current = num
            length = 1
            while current + 1 in numbers:
                current += 1
                length += 1
            if length > maximum:
                maximum = length
    return maximum
a = list(map(int, input("Enter packet IDs: ").split()))
print("Longest consecutive sequence:", longest_consecutive(a))
```

```python
# Employee Skill Grouping
def group_skills(a):
    groups = {}
    for word in a:
        key = ''.join(sorted(word))
        if key not in groups:
            groups[key] = []
        groups[key].append(word)
    return list(groups.values())
a = input("Enter skill codes: ").split()
print("Skill groups:", group_skills(a))
```
